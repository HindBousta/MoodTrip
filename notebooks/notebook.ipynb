```python
import sys
from pathlib import Path

# Add the project root directory to Python path
root_path = Path.cwd().parent  # goes one level up from notebooks/
sys.path.append(str(root_path))
```

1- Mood interpretation

```python
from src.llm.mood_interpreter import MoodInterpreter

# 1. Create interpreter instance
interpreter = MoodInterpreter(method="llm_based")

# 2. User input
user_input = "I want a romantic place to go with my partner."

# 3. Convert user input to structured mood JSON
mood_json = interpreter.interpret(user_input)
print("Mood JSON:")
print(mood_json)
```

2- Retrieve Places

```python
from src.retrieval.retrieve import retrieve_places

# 4. Use mood info to query places
query = f"{mood_json['mood']} {', '.join(mood_json['desired_tags'])}"
places_df = retrieve_places(query=query, top_k=3)

# 5. Convert DataFrame to list of dicts for generator
places = places_df.to_dict(orient="records")
print("\nRetrieved Places:")
for p in places:
    print(p["name"], "-", p["tags"])
```

3- Generate Personalized recommendations

```python
from src.llm.generator import generate_personalized_recommendations

# 6. Generate personalized recommendations
recommendations = generate_personalized_recommendations(
    mood_json=mood_json,
    places=places
)

# 7. Display recommendations
print("\nPersonalized Recommendations:")
for rec in recommendations:
    print(f"\nPlace: {rec['place_name']}")
    print(f"Summary: {rec['summary']}")
    print(f"Micro Itinerary: {rec['micro_itinerary']}")
    print(f"Suggested Plan: {rec.get('suggested_plan', 'N/A')}")
```
